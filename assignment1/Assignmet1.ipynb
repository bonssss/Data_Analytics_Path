```python
import pandas as pd
```

```python
def filter_dataframe(df,column_name,filter_value):
    return df[df[column_name] == filter_value]
```

```python
df = pd.read_csv("students.csv")
```

```python
df
```

```python
sanfransisco_students=filter_dataframe(df,"City","San Francisco")
print("Numberof stundents from San fransisco:" , len(sanfransisco_students))
```

```python
chicago_students = filter_dataframe(df, "City", "Chicago")
average_age = chicago_students["Age"].mean()
print("Average age of students from Chicago:", average_age)
```

```python
# 3. Students who are 30 years old or older
students_30_or_older = df[df["Age"] >= 30]
print("\nStudents who are 30 years old or older:")
print(students_30_or_older)

print("Number of students aged 30 or older:", len(students_30_or_older))
```
